```sql
CREATE CATALOG IF NOT EXISTS turbines;
CREATE SCHEMA IF NOT EXISTS turbines.bronze;
CREATE SCHEMA IF NOT EXISTS turbines.silver;
CREATE SCHEMA IF NOT EXISTS turbines.gold;
CREATE SCHEMA IF NOT EXISTS turbines.control;


CREATE TABLE IF NOT EXISTS turbines.control.watermarks (
  source        STRING,
  watermark_ts  TIMESTAMP,
  updated_at    TIMESTAMP
);

CREATE TABLE IF NOT EXISTS turbines.control.pipeline_log (
  run_id      STRING,
  layer       STRING,
  started_at  TIMESTAMP,
  finished_at TIMESTAMP,
  rows_in     LONG,
  rows_out    LONG,
  status      STRING,
  message     STRING
);

CREATE TABLE IF NOT EXISTS turbines.control.error_log (
  run_id    STRING,
  layer     STRING,
  logged_at TIMESTAMP,
  error     STRING
);
```
